```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate a dataset for 50 days
np.random.seed(42)

dates = pd.date_range(start='2026-01-01', periods=50, freq='D')
temperatures = np.random.randint(20, 41, size=50)

# 2. Create a DataFrame
df = pd.DataFrame({
    'Date': dates,
    'Temperature (°C)': temperatures
})

print("Daily Temperature Dataset:")
print(df)

# 3. Create a Line Chart
plt.figure(figsize=(12, 5))

plt.plot(
    df['Date'],
    df['Temperature (°C)'],
    color='blue',
    marker='o',
    linestyle='-',
    linewidth=2,
    markersize=5,
    label='Daily Temperature'
)

plt.title('Daily Temperature Over 50 Days', fontsize=16, color='darkblue')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()

# 4. Create a Scatter Plot
plt.figure(figsize=(12, 5))

plt.scatter(
    df['Date'],
    df['Temperature (°C)'],
    color='red',
    marker='o',
    s=60,
    edgecolors='black',
    alpha=0.8,
    label='Daily Temperature'
)

plt.title('Daily Temperature Scatter Plot', fontsize=16, color='darkred')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris

# 1. Load the Iris dataset
iris = load_iris()

# 2. Create a DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['Species'] = iris.target

# 3. Draw a scatter plot comparing sepal length and sepal width
colors = ['red', 'green', 'blue']

for i, species in enumerate(iris.target_names):
    subset = df[df['Species'] == i]
    plt.scatter(
        subset['sepal length (cm)'],
        subset['sepal width (cm)'],
        color=colors[i],
        label=species,
        s=50,
        alpha=0.8
    )

# 4. Add title and labels
plt.title('Scatter Plot of Iris Dataset', fontsize=14)
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')

# 5. Add legend and grid
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# 6. Display the plot
plt.show()
```
