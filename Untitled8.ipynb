```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# 1. Generate a sample dataset
X, y = make_blobs(
    n_samples=300,
    centers=3,
    n_features=2,
    cluster_std=1.0,
    random_state=42
)

# 2. Plot the original data points
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], color='blue', s=40)

plt.title('Original Dataset')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 3. Apply K-Means clustering (K=3)
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(X)

# Get cluster labels and centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# 4. Visualize clustered data and centroids
plt.figure(figsize=(8, 5))

colors = ['red', 'green', 'blue']

for i in range(3):
    plt.scatter(
        X[labels == i, 0],
        X[labels == i, 1],
        color=colors[i],
        s=40,
        label=f'Cluster {i + 1}'
    )

# Plot centroids
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    color='yellow',
    marker='X',
    s=250,
    edgecolors='black',
    linewidths=1.5,
    label='Centroids'
)

plt.title('K-Means Clustering (K=3)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 5. Display centroid coordinates
print("Cluster Centroids:")
print(centroids)

# Display number of points in each cluster
print("\nNumber of points in each cluster:")
for i in range(3):
    print(f"Cluster {i + 1}: {sum(labels == i)} points")
```
