```python
import pandas as pd
import numpy as np

# 1. Create the employee DataFrame
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 106, 108],
    'Age': [25, 30, np.nan, 45, 28, 150, 150, 35],
    'Department': ['HR', 'IT', 'Finance', 'IT', np.nan, 'HR', 'HR', 'Finance'],
    'Experience': [2, 5, 3, np.nan, 4, 25, 25, 6],
    'Performance_Score': [85, 90, 78, 95, np.nan, 88, 88, 200]
}

employee = pd.DataFrame(data)

print("Original Employee DataFrame:")
print(employee)

# 2. Handle missing values
# Fill numerical missing values with the median
numeric_cols = ['Age', 'Experience', 'Performance_Score']

for col in numeric_cols:
    employee[col] = employee[col].fillna(employee[col].median())

# Fill missing Department with the mode
employee['Department'] = employee['Department'].fillna(
    employee['Department'].mode()[0]
)

print("\nAfter Handling Missing Values:")
print(employee)

# 3. Remove duplicate records
employee = employee.drop_duplicates()

print("\nAfter Removing Duplicates:")
print(employee)

# 4. Detect outliers using the IQR method
# Apply IQR to numerical columns
outlier_summary = {}

for col in numeric_cols:
    Q1 = employee[col].quantile(0.25)
    Q3 = employee[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = employee[
        (employee[col] < lower_bound) |
        (employee[col] > upper_bound)
    ]

    outlier_summary[col] = outliers[col].tolist()

    print(f"\nOutliers in {col}:")
    print(outliers[['Employee_ID', col]])

    # Replace outliers with the corresponding boundary values
    employee[col] = employee[col].clip(
        lower=lower_bound, upper=upper_bound
    )

print("\nOutlier Summary:")
print(outlier_summary)

# 5. Display the final cleaned DataFrame
print("\nFinal Cleaned Employee DataFrame:")
print(employee)

# 6. Check for remaining missing values and duplicates
print("\nMissing Values:")
print(employee.isnull().sum())

print("\nNumber of Duplicate Rows:", employee.duplicated().sum())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Load the Iris dataset
iris = load_iris()

# 2. Create a DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df['Species'] = iris.target

print("Original Iris DataFrame:")
print(df.head())

# 3. Separate features and target
X = df[iris.feature_names]
y = df['Species']

# 4. Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 5. Apply PCA to reduce 4D data to 2D
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# 6. Create a new DataFrame with 2 principal components
pca_df = pd.DataFrame(
    X_pca,
    columns=['Principal Component 1', 'Principal Component 2']
)

pca_df['Species'] = y

print("\nDataFrame after PCA (2D):")
print(pca_df.head())

# 7. Display explained variance
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

print(
    "\nTotal Variance Retained:",
    round(sum(pca.explained_variance_ratio_) * 100, 2), "%"
)

# 8. Visualize the 2D data
colors = ['red', 'green', 'blue']
species_names = iris.target_names

plt.figure(figsize=(8, 6))

for i, species in enumerate(species_names):
    plt.scatter(
        pca_df.loc[pca_df['Species'] == i, 'Principal Component 1'],
        pca_df.loc[pca_df['Species'] == i, 'Principal Component 2'],
        color=colors[i],
        label=species
    )

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Iris Dataset: PCA (4D to 2D)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA

# 1. Create Customer Profile Dataset
profile = pd.DataFrame({
    'Customer_ID': [1, 2, 3, 4, 5, 6],
    'Age': [25, 35, np.nan, 45, 30, 50],
    'Gender': ['Male', 'Female', 'Male', np.nan, 'Female', 'Male'],
    'Income': [30000, 60000, 45000, 80000, np.nan, 90000]
})

# 2. Create Purchase History Dataset
purchase = pd.DataFrame({
    'Customer_ID': [1, 2, 3, 4, 5, 6],
    'Total_Purchases': [5, 10, 7, 15, np.nan, 20],
    'Total_Spending': [2000, 8000, 4000, 15000, 6000, 25000]
})

print("Customer Profile Dataset:")
print(profile)

print("\nPurchase History Dataset:")
print(purchase)

# 3. Data Integration
df = pd.merge(profile, purchase, on='Customer_ID', how='inner')

print("\nIntegrated Dataset:")
print(df)

# 4. Data Cleaning: Handle Missing Values
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Income'] = df['Income'].fillna(df['Income'].median())
df['Total_Purchases'] = df['Total_Purchases'].fillna(
    df['Total_Purchases'].median()
)
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

df = df.drop_duplicates()

print("\nCleaned Dataset:")
print(df)

# 5. Data Transformation: Encode Gender
df['Gender'] = df['Gender'].map({
    'Male': 1,
    'Female': 0
})

# Normalize numerical features
features = [
    'Age', 'Gender', 'Income',
    'Total_Purchases', 'Total_Spending'
]

scaler = MinMaxScaler()
df[features] = scaler.fit_transform(df[features])

print("\nTransformed Dataset (Normalized):")
print(df)

# 6. Data Reduction: Apply PCA
pca = PCA(n_components=2)
reduced_data = pca.fit_transform(df[features])

reduced_df = pd.DataFrame(
    reduced_data,
    columns=['PC1', 'PC2']
)

reduced_df['Customer_ID'] = df['Customer_ID'].values

print("\nReduced Dataset using PCA:")
print(reduced_df)

# 7. Discretization: Classify Spending
# Use original spending values for meaningful thresholds
spending = purchase.set_index('Customer_ID')['Total_Spending']
df['Total_Spending_Original'] = df['Customer_ID'].map(spending)

threshold = df['Total_Spending_Original'].median()

df['Spending_Category'] = np.where(
    df['Total_Spending_Original'] >= threshold,
    'High Spender',
    'Low Spender'
)

print("\nFinal Dataset with Spending Categories:")
print(df[['Customer_ID', 'Total_Spending_Original',
          'Spending_Category']])

# 8. Display PCA variance retained
print("\nExplained Variance Ratio:", pca.explained_variance_ratio_)
print("Total Variance Retained:",
      round(pca.explained_variance_ratio_.sum() * 100, 2), "%")
```
