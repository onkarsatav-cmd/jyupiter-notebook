```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the Auto MPG dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"
df = pd.read_csv(url)

# Select the required numerical attributes
df = df[['mpg', 'horsepower', 'weight',
         'cylinders', 'acceleration']]

# Convert columns to numeric and handle missing values
df = df.apply(pd.to_numeric, errors='coerce')

print("First five rows of the dataset:")
print(df.head())

print("\nMissing values:")
print(df.isnull().sum())

# 2. Handle missing values
df = df.dropna()

# 3. Create box plots for MPG and Horsepower
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['mpg'], color='skyblue')
plt.title('Box Plot of MPG')
plt.ylabel('Miles Per Gallon')

plt.subplot(1, 2, 2)
sns.boxplot(y=df['horsepower'], color='orange')
plt.title('Box Plot of Horsepower')
plt.ylabel('Horsepower')

plt.tight_layout()
plt.show()

# 4. Identify outliers using the IQR method
for col in ['mpg', 'horsepower']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\nOutliers in {col}:")
    print(outliers[col].tolist())
    print("Number of outliers:", len(outliers))

# 5. Generate a correlation heat map
correlation = df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heat Map of Auto MPG Dataset')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create a dataset of daily sales for 2 months (60 days)
sales = [
    1200, 1500, 1800, 1400, 2000, 1700, 1600, 1900, 2200, 1500,
    1300, 2100, 1800, 1600, 2400, 2000, 1700, 1900, 1500, 1800,
    2200, 1600, 1400, 2100, 2500, 1900, 1700, 1800, 2000, 1600,
    1400, 2300, 1800, 1500, 2100, 1700, 1900, 2200, 1600, 1800,
    2000, 1500, 1700, 2400, 1900, 2100, 1800, 1600, 2000, 1700,
    1500, 2200, 1800, 1900, 1600, 2100, 1700, 2500, 1800, 10000
]

# 2. Create a DataFrame
df = pd.DataFrame({
    'Day': range(1, 61),
    'Daily_Sales': sales
})

print("Retail Daily Sales Dataset:")
print(df)

# 3. Calculate outliers using the IQR method
Q1 = df['Daily_Sales'].quantile(0.25)
Q3 = df['Daily_Sales'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['Daily_Sales'] < lower_bound) |
    (df['Daily_Sales'] > upper_bound)
]

print("\nLower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("\nOutliers:")
print(outliers)

# 4. Generate the Box Plot
plt.figure(figsize=(8, 6))

sns.boxplot(
    y=df['Daily_Sales'],
    color='skyblue',
    width=0.4,
    flierprops={
        'marker': 'o',
        'markerfacecolor': 'red',
        'markersize': 8
    }
)

plt.title('Box Plot of Daily Retail Sales (60 Days)',
          fontsize=14, color='darkblue')
plt.ylabel('Daily Sales (₹)', fontsize=12)
plt.xlabel('Retail Store', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
